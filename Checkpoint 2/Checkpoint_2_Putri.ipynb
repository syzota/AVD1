## Outliers

Outlier adalah nilai yang sangat berbeda dari nilai lainnya dalam dataset. Nilai ini bisa lebih rendah atau lebih tinggi. Mereka dapat muncul karena berbagai alasan, seperti faktor kesalahan atau kejadian lain yang tidak terduga.

```python
results = []

cols = df.select_dtypes(include=['float64', 'int64'])

for col in cols:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1
  lower_bound = q1 - 1.5*iqr
  upper_bound = q3 + 1.5*iqr
  outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
  percent_outliers = (len(outliers)/len(df))*100
  results.append({'Kolom': col, 'Persentase Outliers': percent_outliers})

results_df = pd.DataFrame(results)
results_df.set_index('Kolom', inplace=True)
results_df = results_df.rename_axis(None, axis=0).rename_axis('Kolom', axis=1)

display(results_df)
```

Presentase *Speechines* dan *Liveness* tidak lebih terlalu tinggi, tapi akan dilakukan imputasi dengan metode **Winsorizing**, penggantian nilai terkecil dan terbesar dari suatu set data dengan yang paling dekat.

```python
columns_to_impute = ["Speechiness", "Liveness"]

for col in columns_to_impute:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df.loc[:, col] = df[col].clip(lower=lower_bound, upper=upper_bound)
```

```python
results = []

cols = df.select_dtypes(include=['float64', 'int64'])

for col in cols:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1
  lower_bound = q1 - 1.5*iqr
  upper_bound = q3 + 1.5*iqr
  outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
  percent_outliers = (len(outliers)/len(df))*100
  results.append({'Kolom': col, 'Persentase Outliers': percent_outliers})

results_df = pd.DataFrame(results)
results_df.set_index('Kolom', inplace=True)
results_df = results_df.rename_axis(None, axis=0).rename_axis('Kolom', axis=1)

display(results_df)
```

# Construct Data

Konstruksi data berarti membangun atau membuat fitur (karakteristik) baru dari data yang ada atau mengubah struktur data agar sesuai dengan kebutuhan.


Dalam kasus ini, kita akan mengelompokkan lagu-lagu ke dalam masing-masing dekade karena kebutuhan bisnis: menganalisa tren musik pada tahun 2000s.

```python

def get_decade(year):
    return f"{(year // 10) * 10}s"

df['Decade'] = df['Year'].apply(get_decade)

print(df[['Year', 'Decade']].head())
```
