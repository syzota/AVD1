Penyajian data dalam bentuk grafis atau visual seperti grafik, diagram, dan peta disebut visualisasi data. Tujuannya adalah untuk membantu orang memahami pola, tren, dan hubungan antara data dengan lebih sederhana.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/drive/MyDrive/Praktikum AVD Archive/filtered_spotify.csv")
df
```

# **Data Kategorikal**

# Bar Chart

Grafik ini cocok untuk membandingkan nilai antara kategori yang berbeda, contoh dalam dataset ini akan diperlihatkan genre musik mana yang paling populer.

```python
top = df['Top Genre'].value_counts().nlargest(20)

plt.figure(figsize = (10, 6))
plt.bar(top.index, top.values, color='#FF2929')
plt.title('20 Genre Musik Terpopuler')
plt.xlabel('Genre Musik')
plt.ylabel('Jumlah Lagu')
plt.xticks(rotation = 80)
plt.show()
```

# Pie Chart

Proporsi atau distribusi suatu kategori dalam bentuk persentase yang bisa ditunjukkan dalam dataset ini adalah kapan lagu-lagu populer ini dirilis.

```python
plt.figure(figsize=(6, 6))
plt.pie(df['Year'].value_counts(), labels=df['Year'].value_counts().index, autopct='%1.1f%%', colors=['#F7F2EB', '#E6F8CC', '#006ADD', '#1AC469', '#EF5FA4',
          '#F26A43', '#30A9E0', '#FFDF00', '#FED3B6', '#C5C1EA'])
plt.title('Tahun Rilis')
plt.show()
```

# Histogram

Di sini akan memperlihatkan kecenderungan data tertentu, contohnya bisa energy, BPM, dancebility, ataupun loudness. Tapi, data yang paling cocok dengan tujuan bisnisnya adalah popularity.

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['Popularity'], bins=20, kde=True, color='#FFDF00')
plt.title('Distribusi Popularitas Lagu di Spotify')
plt.xlabel('Popularitas')
plt.ylabel('Frekuensi')
plt.show()
```

# Boxplot

Kita akan melihat outlier, distribusi,
serta perbedaan antar kategori dalam kolom tempo lagu atau BPM dengan genre.

```python
top_10 = df['Top Genre'].value_counts().index[:10]
df_top_genres = df[df['Top Genre'].isin(top_10)]

plt.figure(figsize=(12, 6))
sns.boxplot(x='Top Genre', y='Beats Per Minute (BPM)', data=df_top_genres, palette='rocket')
plt.title('Distribusi BPM Berdasarkan 10 Genre Teratas')
plt.xlabel('Genre')
plt.ylabel('Beats Per Minute (BPM)')
plt.xticks(rotation=45)
plt.show()
```

# **Hubungan Antar Variabel**

# Scatter Plot

Grafik ini digunakan untuk menampilkan hubungan 2 numerik, dalam hal ini kita akan mencari tahu hubungan durasi dengan popularitas.

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['Popularity'], df['Length (Duration)'], c='blueviolet', alpha=0.5)
plt.title('Popularitas vs Durasi Lagu')
plt.xlabel('Popularitas')
plt.ylabel('Durasi (detik)')
plt.show()
```

# Heatmap

Grafik ini sangat cocok untuk menunjukkan korelasi secara lebih dari 2 atau bahkan keseluruhan. Grafik ini juga sebenarnya versi visualisasi dari sintaks 'df.corr'.

```python
plt.figure(figsize=(8, 5))
sns.heatmap(df[['Beats Per Minute (BPM)', 'Energy', 'Danceability', 'Loudness (dB)', 'Valence', 'Popularity']].corr(), annot=True, cmap='Spectral', fmt='.2f')
plt.title('Heatmap Korelasi Antar Elemen')
plt.show()
```
